```python
import numpy as np
import torch
import h5py
import drwho.data as dat
```

```python
DATAFILE  = 'data/5x64x64_testing_with_morphology.hdf5'

if not dat.download(DATAFILE):
    raise RuntimeError(f'''
    Download of file {datafile} failed!
    ''')
    
def load_images(filename=DATAFILE):
    with h5py.File(filename, 'r') as F:
        image = np.asarray(F['image'][:]).astype(np.float32)
        z     = np.asarray(F['specz_redshift'][:]).astype(np.float32)
        zerr  = np.asarray(F['specz_redshift_err'][:]).astype(np.float32)
    return image, z, zerr

print('reading...')
image, z, zerr = load_images()

print(image.shape)

print('saving...')
N = 25_500
torch.save(
    {
    'image': torch.from_numpy(image[:N]),
    'z':     torch.from_numpy(z)[:N],
    'zerr':  torch.tensor(zerr)[:N]},  
    'data/5x64x64GalaxiesML.pth')
print('\n\tdone!')
```
